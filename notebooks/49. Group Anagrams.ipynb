# 49. Group Anagrams

Difficulty: Medium

Given an array of strings strs, group the

together. You can return the answer in any order.

## Examples

Example 1:

    Input: strs = ["eat","tea","tan","ate","nat","bat"]
    
    Output: [["bat"],["nat","tan"],["ate","eat","tea"]]

Explanation:

    There is no string in strs that can be rearranged to form "bat".
    The strings "nat" and "tan" are anagrams as they can be rearranged to form each other.
    The strings "ate", "eat", and "tea" are anagrams as they can be rearranged to form each other.

Example 2:

    Input: strs = [""]
    
    Output: [[""]]

Example 3:

    Input: strs = ["a"]
    
    Output: [["a"]]

## Constraints

- 1 <= strs.length <= 104
- 0 <= strs[i].length <= 100
- strs[i] consists of lowercase English letters.

## Hash Table

### Solution 1

Intuitive:
1. All anagrams, when sorted, have the same string.
2. The sorted string become the key
3. Group all of them using a hash table
4. Finally, dumps all the values of the hash table into result

Submission link: https://leetcode.com/problems/group-anagrams/submissions/1842374138/

```python
from typing import List

class Solution:
    def groupAnagrams(self, strs: List[str]) -> List[List[str]]:
        result = []
        storage = dict()
        for string in strs:
            key = ''.join(sorted(string))
            if not key in storage:
                storage[key] = [string]
            else:
                storage[key].append(string)

        for _, values in storage.items():
            result.append(values)

        return result
```

```python
class Solution:
    def groupAnagrams(self, strs: List[str]) -> List[List[str]]:
        def get_counts(st: str):
            dics = dict()
            for ch in st:
                if ch in dics:
                    dics[ch] += 1
                else:
                    dics[ch] = 1
            import json
            return json.dumps(dics, sort_keys=True)
        
        tries = [(get_counts(i), i) for i in strs]
        results = dict()

        for trie in tries:
            if trie[0] in results:
                results[trie[0]].append(trie[1])
            else:
                results[trie[0]] = [trie[1]]

        return list(results.values())
```

```python
def assert_lists_equal_ignore_order(actual: List[List[str]], expected: List[List[str]]):
    """
    Assert that two lists of lists are equal, ignoring order at both levels.
    
    Args:
        actual: The actual list of lists
        expected: The expected list of lists
    """
    # Convert inner lists to sorted tuples for comparison
    actual_sorted = sorted([tuple(sorted(inner)) for inner in actual])
    expected_sorted = sorted([tuple(sorted(inner)) for inner in expected])
    
    assert actual_sorted == expected_sorted, \
        f"Lists don't match:\nActual:   {actual}\nExpected: {expected}"
```
